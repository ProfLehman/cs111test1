##**Comments**

Comments are notes for humans that tell others essential details about your code, such as **file name**, **date**, **author**, and **description**. They also are used to highlight key code sections, such as **input**, **processing**, and **output**.

**Single-line** comments start with hashtag #.  Python ignores all code after the hash tag.

```
# single-line comment
```

**Multi-line** comments begin with three single quote marks and end with three single quote marks. Python ignores all code between the three single quote marks. The single quote key is found under the double quote key to the left of enter or return key on your keyboard.

```
'''
multi-line comment line one
line two
line three
line four
'''
```

Most coding environments will use a different color to denote single-line and multi-line comments.  With Google colab single comments are shown in green and multi-line comments are shown in red. Either can be used.


````python
# single line comment approach

# file: years.py
# date: 11.1.2024
# author: Norman Forester
# Description: Python comments, print, and variables

'''
multi-line comment approach

file: years.py
date: 11.1.2024
author: Norman Forester
Description: Python comments, print, and variables
'''
````
